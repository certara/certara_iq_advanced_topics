# Advanced tutorial: Render a templated ReactionModel

This notebook renders a four-compartment monospecific anti-ligand model from a pre-built **templated** ReactionModel file using `TemplatedReactionModel`.

The templated file and rendered output are the same as those produced in `Build_Template.ipynb`; structural verification against `model.txt` is not repeated here. This workflow represents the typical production use case: rendering a templated model that has no handwritten reference to compare against.

```python
from pathlib import Path
from utilities import TemplatedReactionModel

TEMPLATED_PATH = Path("inputs/Templated_Model/advanced_templated.model")
RENDERED_OUTPUT_PATH = Path("advanced_rendered.model")
```

## Render with repository tooling

`TemplatedReactionModel.from_text()` reads and parses the templated DSL file. `to_reaction_model()` expands all templates and writes a concrete ReactionModel file.

The rendered output can be exported and saved to the project file system for use as a downstream model input, or this notebook can be declared as a pipeline dependency so the rendered model is regenerated automatically when the template changes.

```python
templated_model = TemplatedReactionModel.from_file(TEMPLATED_PATH)
if not isinstance(templated_model, TemplatedReactionModel):
    raise ValueError(templated_model)

templated_model.to_reaction_model(RENDERED_OUTPUT_PATH)
```
