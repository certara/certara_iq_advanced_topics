# Advanced tutorial: Render a templated ReactionModel

This notebook renders a four-compartment monospecific anti-ligand model imported as a **templated** ReactionModel file. The template is rendered with `TemplatedReactionModel`, and compared against `model.txt`.

The templated model and the final rendered model are identical to those used in the Build_Template.ipynb notebook. So the result is not compared to the reference DSL. This comparison can be found in the Build_Template.ipynb notebook.

For models that are built directly as a templated model and with no reference, the below workflow can be followed.

```python
from pathlib import Path
from utilities import TemplatedReactionModel

TEMPLATED_PATH = Path("inputs/Templated_Model/advanced_templated.model")
RENDERED_OUTPUT_PATH = Path("advanced_rendered.model")
```

## Render with repository tooling

`TemplatedReactionModel.from_text()` parses the templated DSL. `to_reaction_model()` expands every template and writes ordinary ReactionModel text.

The output of this model can be either exported to the project to be used as an input. Or this notebook can be imported as a dependency directly when the rendered model is desired

```python
templated_model = TemplatedReactionModel.from_file(TEMPLATED_PATH)
if not isinstance(templated_model, TemplatedReactionModel):
    raise ValueError(templated_model)

templated_model.to_reaction_model(RENDERED_OUTPUT_PATH)
```
